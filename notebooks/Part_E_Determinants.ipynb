```python
# Import NumPy and linear-algebra utilities for determinant calculations.
import numpy as np
from numpy import linalg
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

```python
# Form X_train^T X_train and calculate its determinant.
XtX_train = X_train.T @ X_train
det_XtX_train = np.linalg.det(XtX_train)

print(f"det(X_train^T X_train) = {det_XtX_train}")
```

```python
# Test invertibility and measure numerical conditioning of the Gram matrix.
# Also compute its condition number for a stability check.
cond_number = np.linalg.cond(XtX_train)

print(f"Condition number = {cond_number}")

if not np.isclose(det_XtX_train, 0):
    print("XtX_train is invertible.")
else:
    print("XtX_train is not invertible.")
```
