```python
# Import numerical, tabular, and plotting libraries used throughout the notebook.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

```python
# Load the public BigO(Bench) data and select the target and Code_Length feature.
from datasets import load_dataset

data = load_dataset(
    "facebook/BigOBench",
    "time_complexity_test_set.jsonl",
    split="train"
).to_pandas()

# Feature matrix: source code length
feature_names = ["Code_Length"]
X_raw = data["solution_code"].str.len().to_numpy().reshape(-1, 1)

# Target vector: time complexity coefficient
y = data["time_curve_coefficient"].to_numpy()

assert X_raw is not None and y is not None, "Load your dataset above before continuing."
n_samples, n_features = X_raw.shape
print(f"n_samples = {n_samples}, n_features = {n_features}")
print(f"feature_names = {feature_names}")
```

```python
# Inspect the dataset shape, columns, and first rows before modeling.
pd.DataFrame(X_raw, columns=feature_names).describe()
```
