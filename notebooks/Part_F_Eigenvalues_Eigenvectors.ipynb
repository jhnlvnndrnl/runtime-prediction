**What to do:** Compute the eigenvalues and eigenvectors of a matrix relevant to your data (e.g., the covariance matrix of your standardized training features).

```python
# Load the transformed training and test arrays saved by Part D.
from google.colab import drive
drive.mount('/content/drive')

import numpy as np
data_d = np.load('/content/drive/MyDrive/part_d_data.npz')
X_train = data_d['X_train']
X_test = data_d['X_test']
y_train = data_d['y_train']
y_test = data_d['y_test']
```

```python
# Import numerical and plotting libraries and set readable output precision.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

```python
# Compute and validate the covariance matrix of the standardized training features.
C = np.atleast_2d(np.cov(X_train, rowvar=False))
print("C shape:", C.shape)
print("Symmetric?", np.allclose(C, C.T))
```

```python
# Compute the symmetric eigendecomposition and order eigenpairs by decreasing eigenvalue.
eigvals, eigvecs = np.linalg.eigh(C)
order = np.argsort(eigvals)[::-1]
eigvals = eigvals[order]
eigvecs = eigvecs[:, order]
print("Eigenvalues (largest to smallest):", eigvals)
```

```python
# Verify the leading eigenpair by evaluating the residual C v - lambda v.
v1, lam1 = eigvecs[:, 0], eigvals[0]
error = np.max(np.abs(C @ v1 - lam1 * v1))
print("max|C v1 - lambda1 v1| =", error)
```

```python
# Calculate variance explained by each principal direction and display the leading loadings.
explained_var_ratio = eigvals / eigvals.sum()
print("Explained variance ratio:", explained_var_ratio)
print("Leading eigenvector loadings:", eigvecs[:, 0])
```
