```python
# Import libraries needed for standardization and data inspection.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

```python
# Build the standardization matrix and apply the centered linear transformation.
# Example: standardization T(x) = S(x - mean), whose LINEAR part is
# multiplication by S = diag(1/std):
mu = X.mean(axis=0)
sigma = X.std(axis=0)
S = np.diag(1 / sigma)

X_centered = X - mu
X_transformed = X_centered @ S
print("X_transformed shape:", None if X_transformed is None else X_transformed.shape)
```

```python
# Demonstrate the transformation on one real sample and verify the matrix calculation.
idx = 0   # pick a sample index
x_before = X[idx]
x_after = X_transformed[idx] if X_transformed is not None else None
print("Before:", x_before)
print("After: ", x_after)
print("Check: ", S @ (x_before - mu))
```

```python
# Randomly split the transformed observations into reproducible training and test sets.
# Example:
# rng = np.random.default_rng(42)
# n = X_transformed.shape[0]
# idx = rng.permutation(n)
# train_size = int(0.8 * n)
# train_idx, test_idx = idx[:train_size], idx[train_size:]
# X_train, X_test = X_transformed[train_idx], X_transformed[test_idx]
# y_train, y_test = y[train_idx], y[test_idx]

rng = np.random.default_rng(42)
n = X_transformed.shape[0]
idx = rng.permutation(n)
train_size = int(0.8 * n)
train_idx, test_idx = idx[:train_size], idx[train_size:]
X_train, X_test = X_transformed[train_idx], X_transformed[test_idx]
y_train, y_test = y[train_idx], y[test_idx]
```

**Q13.** Which linear transformation did you apply to your data? Write its matrix representation M such that T(x) = Mx.

> *Your answer:* I standardized each feature using the training-set mean and standard deviation. Its linear matrix part is $S=\operatorname{diag}(1/\sigma)$, and the complete transformation is $T(x)=S(x-\mu)$ (implemented row-wise as $(x-\mu)S$). With one feature, $S$ is a $1\times1$ matrix containing $1/\sigma$.

**Q14.** Using one actual data point from your dataset, show the before vector x and the after vector T(x) = Mx.

> *Your answer:* For the selected sample `idx = 0`, the notebook prints the original vector `x_before` and its standardized vector `x_after`. In symbols, if the raw value is `x_before[0]`, then $T(x)= (x_{before}[0]-\mu)/\sigma$; the printed `Check` confirms the same result using the matrix representation.

**Q15.** Is your transformation invertible? How do you know, and why would invertibility matter if you wanted to convert a prediction back to original units?

> *Your answer:* Yes, the transformation is invertible as long as the training standard deviation is nonzero. Its inverse is $x=S^{-1}T(x)+\mu$. This matters because a prediction or feature value in standardized units can be converted back to the original units for interpretation.

**Q16.** Would your transformation's matrix change if computed from a different subset of your data (e.g., a different train/test split)? What does that imply about how you must apply it consistently to both training and test data?

> *Your answer:* Yes. A different subset can have a different mean and standard deviation, so it can produce a different matrix $S$. The parameters must therefore be computed using the training set only, then reused unchanged for both training and test data to prevent data leakage and keep the feature scale consistent.
