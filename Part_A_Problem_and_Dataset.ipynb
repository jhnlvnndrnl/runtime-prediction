**Q1.** What real-world quantity are you trying to predict, and why is it continuous rather than categorical?

**Q2.** Where does your dataset come from? Is it open/public? Provide the source (link or citation).

> BigO(Bench) is a publicly available dataset on Hugging Face (BigO(Bench)-Hugging Face). The group uses only its time_complexity_test_set subset, which contains 640 human-written Python solutions to 311 competitive-programming problems, each with public, private, and generated test inputs and expected outputs.

> PCMRD is student-created because every value in it is counted or measured by the group itself on its own computer. The Python programs and test inputs come from BigO(Bench), but BigO(Bench)’s precomputed runtimes and complexity labels are not copied and are not used as model variables.

> The data collection method is measurement. LOC and CC are obtained by counting from the source code, and execution time is obtained by timing actual program runs. Each program is run once untimed as a warm-up and then 5 times timed, and the mean of the 5 timed runs is recorded as the execution time. A run is valid only if the program’s output matches the expected output and it finishes within 10 seconds.

> The dataset is stored as a CSV file that is loaded with Python using pandas.

> Citation: Chambon et al. (2025). BigO(Bench). Hugging Face. https://huggingface.co/datasets/facebook/BigOBench



**Q3.** How many samples (n) and how many features (p) does your dataset have?

**Q4.** Before doing any analysis, which features do you expect to be most predictive, and why?

> The project expects Lines of Code (LOC), Cyclomatic Complexity (CC), and Input Size to influence software execution time.

1. Lines of Code (LOC) - it is the number of lines of text in a computer program's source code, ignoring blank lines and comments. Execution times may take longer for programs having more statement lines.

2. Cyclomatic Complexity (CC) - is the quantitative metric of the number of independent paths through a program’s control flow graph. Execution times may take longer because of many decision points and loops having multiple execution paths.

3. Input Size - the amount of input processed in the running of the computer program. Larger input size requires the program to perform more, taking longer for execution.

> Although the above listed independent variables are predictors, no single variable can be said to solely determine execution time. Algorithms, hardware, and software environment are some other factors that execution time may depend on. Statistical relationship would be examined for the variables and execution time.
